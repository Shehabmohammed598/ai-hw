```python
# تثبيت المكتبات الضرورية
!pip install flask pyngrok --quiet

# استيراد المكتبات
from flask import Flask, render_template_string, request, redirect, url_for
from pyngrok import ngrok

# إنشاء تطبيق Flask
app = Flask(__name__)

# صفحة تسجيل الدخول
login_page = """
<!DOCTYPE html>
<html>
<head>
    <title>تسجيل الدخول</title>
</head>
<body>
    <h1>تسجيل الدخول</h1>
    <form action="/login" method="post">
        <label>اسم المستخدم:</label>
        <input type="text" name="username" required><br>
        <label>كلمة المرور:</label>
        <input type="password" name="password" required><br>
        <button type="submit">تسجيل الدخول</button>
    </form>
    <a href="/signup">إنشاء حساب جديد</a>
</body>
</html>
"""

# صفحة إنشاء حساب
signup_page = """
<!DOCTYPE html>
<html>
<head>
    <title>إنشاء حساب</title>
</head>
<body>
    <h1>إنشاء حساب جديد</h1>
    <form action="/signup" method="post">
        <label>اسم المستخدم:</label>
        <input type="text" name="username" required><br>
        <label>كلمة المرور:</label>
        <input type="password" name="password" required><br>
        <button type="submit">إنشاء حساب</button>
    </form>
</body>
</html>
"""

# صفحة المنتجات
products_page = """
<!DOCTYPE html>
<html>
<head>
    <title>المنتجات</title>
</head>
<body>
    <h1>المنتجات</h1>
    <h3>اختر فئة:</h3>
    <a href="/products/men">رجالي</a><br>
    <a href="/products/women">نسائي</a><br>
    <a href="/products/kids">أطفال</a><br>
    <a href="/products/fragrances">عطور</a>
</body>
</html>
"""

# صفحة الفئة المحددة
category_page_template = """
<!DOCTYPE html>
<html>
<head>
    <title>المنتجات - {{ category }}</title>
</head>
<body>
    <h1>المنتجات - {{ category }}</h1>
    <ul>
        {% for product in products %}
        <li>{{ product['name'] }} - {{ product['price'] }} ريال {% if product['discount'] %}(خصم!) {% endif %}</li>
        {% endfor %}
    </ul>
    <a href="/products">عودة إلى الفئات</a>
</body>
</html>
"""

# الصفحات
@app.route('/')
def home():
    return redirect('/login')

@app.route('/login', methods=['GET', 'POST'])
def login():
    if request.method == 'POST':
        username = request.form['username']
        password = request.form['password']
        # هنا يتم التحقق من بيانات تسجيل الدخول
        return redirect('/products')
    return render_template_string(login_page)

@app.route('/signup', methods=['GET', 'POST'])
def signup():
    if request.method == 'POST':
        username = request.form['username']
        password = request.form['password']
        # هنا يتم تخزين بيانات المستخدم الجديد
        return redirect('/login')
    return render_template_string(signup_page)

@app.route('/products', methods=['GET'])
def products():
    return render_template_string(products_page)

@app.route('/products/<category>', methods=['GET'])
def category_products(category):
    # بيانات تجريبية
    products = [
        {"name": "منتج 1", "price": 100, "discount": False},
        {"name": "منتج 2", "price": 150, "discount": True},
        {"name": "منتج 3", "price": 200, "discount": False},
    ]
    return render_template_string(category_page_template, category=category, products=products)

# تكامل مع ngrok لتوفير رابط عام
public_url = ngrok.connect(5000)
print("Public URL:", public_url)

# تشغيل التطبيق
if __name__ == '__main__':
    app.run(port=5000)
```
